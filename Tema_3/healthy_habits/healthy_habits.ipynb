Programa interactivo para evaluar hábitos saludables
y proporcionar recomendaciones generales.

```python
# Aquí empezará el programa
cabecera_1 = r"""{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}
{} __  __     ______     ______     __         ______   __  __     __  __   {}
{}/\ \_\ \   /\  ___\   /\  __ \   /\ \       /\__  _\ /\ \_\ \   /\ \_\ \  {}
{}\ \  __ \  \ \  __\   \ \  __ \  \ \ \____  \/_/\ \/ \ \  __ \  \ \____ \ {}
{} \ \_\ \_\  \ \_____\  \ \_\ \_\  \ \_____\    \ \_\  \ \_\ \_\  \/\_____\{}
{}  \/_/\/_/   \/_____/   \/_/\/_/   \/_____/     \/_/   \/_/\/_/   \/_____/{}
{}                                                                          {}
{} __  __     ______     ______     __     ______   ______                  {}
{}/\ \_\ \   /\  __ \   /\  == \   /\ \   /\__  _\ /\  ___\                 {}
{}\ \  __ \  \ \  __ \  \ \  __<   \ \ \  \/_/\ \/ \ \___  \                {}
{} \ \_\ \_\  \ \_\ \_\  \ \_____\  \ \_\    \ \_\  \/\_____\               {}
{}  \/_/\/_/   \/_/\/_/   \/_____/   \/_/     \/_/   \/_____/               {}
{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}{}"""


def obtener_datos():
    nombre = input("\n👤 ¿Cuál es tu nombre? ")

    while True:
        try:
            edad = int(input("🎂 ¿Qué edad tienes? "))

            if edad >= 14 and edad <= 100:
                print("Edad:", edad)
                break
            else:
                print("La edad debe estar entre 14 y 100 años")

        except ValueError:
            print("Es necesario introducir un número entero")

    while True:
        try:
            peso = float(input("⚖️  ¿Cuánto pesas? (kg): "))

            if peso >= 30 and peso <= 300:
                print("Peso:", peso)
                break
            else:
                print("El peso debe estar entre 30 y 300 kg")

        except ValueError:
            print("Es necesario introducir un número")

    while True:
        try:
            altura = float(input("📏 ¿Cuánto mides? (metros, ej.: 1.75): "))

            if altura >= 1.20 and altura <= 2.30:
                print("Altura:", altura)
                break
            else:
                print("La altura debe estar entre 1.20 y 2.30 metros")

        except ValueError:
            print("Es necesario introducir un número")

    sexo = input("⚤  ¿Eres hombre o mujer? (H/M): ").strip().lower()

    while sexo not in ("h", "hombre", "m", "mujer"):
        print("Respuesta no válida")
        sexo = input("Introduce H o M: ").strip().lower()

    return nombre, edad, peso, altura, sexo
```

```python
def preguntar_si_no(pregunta):
    respuesta = input(pregunta + " (S/N): ").strip().lower()

    while respuesta not in ("s", "sí", "si", "n", "no"):
        print("Respuesta no válida")
        respuesta = input("Introduce sí o no: ").strip().lower()

    return respuesta


def cuestionario_bucodental():
    dolor_dental = preguntar_si_no(
        "¿Tienes dolor dental actualmente?"
    )
    sangrado_encias = preguntar_si_no(
        "¿Te sangran las encías con frecuencia?"
    )
    sensibilidad_dental = preguntar_si_no(
        "¿Tienes sensibilidad dental frecuente al frío, calor o alimentos dulces?"
    )
    llagas_persistentes = preguntar_si_no(
        "¿Tienes alguna llaga o lesión en la boca que lleve más de dos semanas sin desaparecer?"
    )
    mal_aliento = preguntar_si_no(
        "¿Tienes mal aliento persistente a pesar de mantener una higiene bucal habitual?"
    )
    dientes_moviles = preguntar_si_no(
        "¿Has notado algún diente permanente flojo o con movilidad?"
    )

    return {
        "dolor_dental": dolor_dental,
        "sangrado_encias": sangrado_encias,
        "sensibilidad_dental": sensibilidad_dental,
        "llagas_persistentes": llagas_persistentes,
        "mal_aliento": mal_aliento,
        "dientes_moviles": dientes_moviles
    }


def higiene_bucodental():

    while True:
        try:
            cepillados_dia = int(
                input("¿Cuántas veces te cepillas al día? ")
            )

            if 0 <= cepillados_dia <= 4:
                break

            print("Introduce un valor entre 0 y 4")

        except ValueError:
            print("Introduce un número entero")

    while True:
        try:
            duracion_cepillado = int(
                input("¿Durante cuántos minutos te cepillas cada vez? ")
            )

            if 0 <= duracion_cepillado <= 10:
                break

            print("Introduce un valor entre 0 y 10")

        except ValueError:
            print("Introduce un número entero")

    usa_fluor = input(
        "¿La pasta de dientes que utilizas contiene flúor? (sí/no/no sé): "
    ).strip().lower()

    while usa_fluor not in (
        "s", "sí", "si", "n", "no", "no sé", "no se"
    ):
        print("Respuesta no válida")
        usa_fluor = input(
            "Introduce sí, no o no sé: "
        ).strip().lower()

    usa_interdental = input(
        "¿Utilizas hilo dental o limpiador interdental? (S/N): "
    ).strip().lower()

    while usa_interdental not in ("s", "sí", "si", "n", "no"):
        print("Respuesta no válida")
        usa_interdental = input(
            "Introduce sí o no: "
        ).strip().lower()

    frecuencia_interdental = None

    if usa_interdental in ("s", "si", "sí"):
        print("""
¿Con qué frecuencia usas hilo dental o limpiador interdental?

1 - Ocasionalmente
2 - Varias veces por semana
3 - Una vez al día
4 - Más de una vez al día
""")

        while True:
            try:
                frecuencia_interdental = int(
                    input("Selecciona una opción (1-4): ")
                )

                if 1 <= frecuencia_interdental <= 4:
                    break

                print("Introduce un valor entre 1 y 4")

            except ValueError:
                print("Debe ser un número entero")

    limpieza_lingual = input(
        "¿Sueles limpiarte la lengua? (S/N): "
    ).strip().lower()

    while limpieza_lingual not in ("s", "sí", "si", "n", "no"):
        print("Respuesta no válida")
        limpieza_lingual = input(
            "Introduce sí o no: "
        ).strip().lower()

    frecuencia_lingual = None

    if limpieza_lingual in ("s", "si", "sí"):
        print("""
¿Con qué frecuencia te limpias la lengua?

1 - Ocasionalmente
2 - Varias veces por semana
3 - Una vez al día
4 - Más de una vez al día
""")

        while True:
            try:
                frecuencia_lingual = int(
                    input("Selecciona una opción (1-4): ")
                )

                if 1 <= frecuencia_lingual <= 4:
                    break

                print("Introduce un valor entre 1 y 4")

            except ValueError:
                print("Debe ser un número entero")

    print("""
¿Con qué frecuencia acudes al dentista para una revisión?

1 - Cada 6 meses o menos
2 - Aproximadamente una vez al año
3 - Cada 2 años
4 - Con menos frecuencia
5 - Solo cuando tengo algún problema
6 - Nunca
""")

    while True:
        try:
            frecuencia_revision = int(
                input("Selecciona una opción (1-6): ")
            )

            if 1 <= frecuencia_revision <= 6:
                break

            print("Introduce un valor entre 1 y 6")

        except ValueError:
            print("Debe ser un número entero")

    problema_bucodental = input(
        "¿Tienes algún problema bucodental? (S/N): "
    ).strip().lower()

    while problema_bucodental not in ("s", "sí", "si", "n", "no"):
        print("Respuesta no válida")
        problema_bucodental = input(
            "Introduce sí o no: "
        ).strip().lower()

    descripcion_problema = None
    quiere_cuestionario = None
    resultados_cuestionario = None

    if problema_bucodental in ("s", "si", "sí"):
        descripcion_problema = input(
            "Describe brevemente el problema: "
        )

    else:
        quiere_cuestionario = input(
            "¿Te gustaría realizar un breve cuestionario para detectar "
            "posibles problemas bucodentales? (S/N): "
        ).strip().lower()

        while quiere_cuestionario not in (
            "s", "sí", "si", "n", "no"
        ):
            print("Respuesta no válida")
            quiere_cuestionario = input(
                "Introduce sí o no: "
            ).strip().lower()

        if quiere_cuestionario in ("s", "si", "sí"):
            resultados_cuestionario = cuestionario_bucodental()

    return {
        "cepillados_dia": cepillados_dia,
        "duracion_cepillado": duracion_cepillado,
        "usa_fluor": usa_fluor,
        "usa_interdental": usa_interdental,
        "frecuencia_interdental": frecuencia_interdental,
        "limpieza_lingual": limpieza_lingual,
        "frecuencia_lingual": frecuencia_lingual,
        "frecuencia_revision": frecuencia_revision,
        "problema_bucodental": problema_bucodental,
        "descripcion_problema": descripcion_problema,
        "quiere_cuestionario": quiere_cuestionario,
        "resultados_cuestionario": resultados_cuestionario
    }
```
